# GPU no Claude + API (texto, imagem, vídeo, fala, transcrição)

1. Ative a GPU: **Ambiente de execução → Alterar tipo → T4 GPU** (no Kaggle: Settings → Accelerator → GPU).
2. Cole seu token do ngrok abaixo (https://dashboard.ngrok.com/get-started/your-authtoken).
3. Rode as células. Aparecem 3 URLs:
   - `conector_claude` → Claude → Configurações → Conectores → Adicionar conector personalizado
   - `api` → API própria (`/texto`, `/imagem`, `/video`, `/fala`, `/transcrever`, `/pipeline`)
   - `openai_base_url` → use em qualquer app/biblioteca compatível com OpenAI

⚠️ As URLs contêm um token secreto: não compartilhe. Confira os termos de uso da plataforma onde você roda este notebook.
