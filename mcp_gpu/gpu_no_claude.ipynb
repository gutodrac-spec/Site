# GPU no Claude via conector MCP

1. Ative a GPU: **Ambiente de execução → Alterar tipo → T4 GPU** (no Kaggle: Settings → Accelerator → GPU).
2. Cole seu token do ngrok abaixo (https://dashboard.ngrok.com/get-started/your-authtoken).
3. Rode as células e copie a URL que aparecer.
4. No Claude: **Configurações → Conectores → Adicionar conector personalizado** e cole a URL.

⚠️ A URL contém um token secreto: não compartilhe. Confira os termos de uso da plataforma onde você roda este notebook.
